Given the following data:

```python
hotel_reservations = [
    {"guest": "Arif", "room_cost": 600000, "reservation_status": "COMPLETED"},
    {"guest": "Bella", "room_cost": 300000, "reservation_status": "PENDING"},
    {"guest": "Chika", "room_cost": 2500000, "reservation_status": "COMPLETED"},
    {"guest": "Doni", "room_cost": "N/A", "reservation_status": "COMPLETED"},
    {"guest": "Ega", "room_cost": 800000, "reservation_status": "COMPLETED"}
]
```

1. Create a generator expression that lazily filters guest records and only contains records whose reservation status is "COMPLETED".

```python
completed_reservations_gen = (reservation for reservation in hotel_reservations if reservation["reservation_status"] == "COMPLETEDc")
```

2. Complete the provided function by filling in the missing parts. The function must:
- Retrieve data manually from the generator
- Prevent side effects by protecting the original data
- Handle invalid data using exception handling

```python
import copy

def process_rebate_pure(reservation_generator):
    """
    desc: Processes reservation data from a generator,
          filters invalid records, and calculates
          the total room cost after rebate deduction.
    """

    total_revenue_after_rebate = 0

    while True:

        try:

            # TODO 1:
            # Manually retrieve the next element from the generator.

            current_reservation = next(reservation_generator)

            clean_cost = int(current_reservation["room_cost"])

            # Rebate Rules:
            # Room cost >= 2,000,000 receives 150,000 rebate
            # Otherwise receives 15,000 rebate

            rebate = (150000 if clean_cost >= 2000000 else 15000)

            # TODO 2:
            # The code below directly modifies the original data
            # (Side Effect). Apply a technique to prevent this issue.

            current_reservation = copy.deepcopy(current_reservation)

            current_reservation["room_cost"] = (
                clean_cost - rebate
            )

            total_revenue_after_rebate += (
                current_reservation["room_cost"]
            )

        # TODO 3:
        """
        Handle invalid reservation data (e.g., "N/A")
        without stopping the program execution.

        Display the guest name whose data is invalid,
        then continue processing the remaining records.
        """

        # TODO 4:
        """
        Handle the condition when the generator has no more data.

        Stop the loop gracefully and terminate the processing.
        """

    return total_revenue_after_rebate
```

3. Create a list comprehension that retrieves guest whose room cost is greater than 500,000.

```python
threshold = 500000

loyal_guests = set(
    reservation["guest"]
    for reservation in hotel_reservations
    if reservation["total_spent"] >= threshold
)
```

4. Display the final results

```python
revenue = process_rebate_pure(completed_reservations_gen) #Data that has been filtered from the generator expression
print(f"Net Revenue: Rp{revenue}")
print(f"Loyal Guests: {loyal_guests}")
```
