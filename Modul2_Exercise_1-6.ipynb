## Exercise 1 — Functional Analysis




```python
#202410370110424
# Data awal
exam_scores = [70, 80, 85]

def add_bonus2(bonus):
    for i in range(len(exam_scores)):
        exam_scores[i] += bonus
    return exam_scores
```

```python
#202410370110424
# Jalankan cell ini 2–3 kali untuk melihat perubahan berulang
print("Before:", exam_scores)
result = add_bonus2(5)
print("New result:", result)
print("Original exam_scores:", exam_scores)
```

## Exercise 2 — Data Duplication with Shallow and Deep Copy



```python
#202410370110424
import copy

user = {"name": "putri", "scores": [90, 85]}

user_shallow = copy.copy(user)
user_deep = copy.deepcopy(user)

# Ubah skor pertama pada kedua salinan
user_shallow["scores"][0] = 75
user_deep["scores"][0] = 75

print("Original user:", user)
print("Shallow copy:", user_shallow)
print("Deep copy:", user_deep)

print("\nID list scores:")
print("user:", id(user["scores"]))
print("user_shallow:", id(user_shallow["scores"]))
print("user_deep:", id(user_deep["scores"]))

print("\nApakah list scores memiliki objek yang sama?")
print("user dan user_shallow:", user["scores"] is user_shallow["scores"])
print("user dan user_deep:", user["scores"] is user_deep["scores"])
```

## Exercise 3 — Using `iter()` and `next()` to Process Tuple Data



```python
#202410370110424
sensor_data = (23.5, 24.0, 22.8, 21.1)

sensor_iterator = iter(sensor_data)

print("Nilai pertama:", next(sensor_iterator))
print("Nilai kedua:", next(sensor_iterator))

print("Data yang tersisa:")
for value in sensor_iterator:
    print(value)
```

## Exercise 4 — Convert to List Comprehension




```python
#202410370110424
word = "universitas"

# Versi awal menggunakan loop
result = []
for letter in word:
    if letter not in "aiueo":
        result.append(letter.upper())

print("Hasil loop:", result)

# Versi List Comprehension
result_comprehension = [
    letter.upper()
    for letter in word
    if letter not in "aiueo"
]

print("Hasil List Comprehension:", result_comprehension)
print("Tipe objek:", type(result_comprehension))
```

### Small Experiment — List Comprehension vs Generator Expression



```python
#202410370110424
numbers_list = [x * 10 for x in range(1, 11) if x % 2 == 0]
numbers_generator = (x * 10 for x in range(1, 11) if x % 2 == 0)

print("List:", numbers_list)
print("Tipe list:", type(numbers_list))
print("Tipe generator:", type(numbers_generator))
print("Nilai generator satu per satu:", list(numbers_generator))
```

## Exercise 5 — Create Generator Expression




```python
#202410370110424
number_generator = (number for number in range(1, 6))

while True:
    try:
        value = next(number_generator)
        print(value)
    except StopIteration:
        print("Iterator telah selesai.")
        break
```

## Exercise 6 — Create Generator Function




```python
#202410370110424
raw_logs = [
    "INFO: User login successful",
    "ERROR: Database connection failed",
    "INFO: Page rendered in 25ms",
    "ERROR: Unauthorized access attempt detected",
    "INFO: User logout successful",
]

def filter_emergency_log(log_list):
    for log in log_list:
        if log.startswith("ERROR"):
            yield log

error_bucket = filter_emergency_log(raw_logs)

print("Tipe error_bucket:", type(error_bucket))
print("Log darurat pertama:", next(error_bucket))
print("Log darurat kedua:", next(error_bucket))
```
