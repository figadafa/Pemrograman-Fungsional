## Part 1 - Strategi Pricing dan Higher-Order Function

```python
# 1
def standard_price(nights, price_per_night):
    return nights * price_per_night

def discount_price(nights, price_per_night):
    return nights * price_per_night * 0.85

def tax_price(nights, price_per_night):
    return nights * price_per_night * 1.10

# 2
def apply_pricing(strategy):
    def add_total(booking):
        # Create a new dictionary to avoid modifying the original booking object directly
        new_booking = booking.copy()
        total_cost = strategy(new_booking["nights"], new_booking["price_per_night"])
        new_booking["total_cost"] = total_cost
        return new_booking
    return add_total

# 3
# You can choose any of the pricing strategies here, e.g., discount_price, standard_price, or tax_price
pricing_strategy = apply_pricing(discount_price)
```

## Part 2 - Predikat untuk Filter

```python
def is_not_cancelled(booking):
    return booking["status"] != "Cancelled"

def is_eligible(booking):
    return booking["nights"] >= 3 and booking["status"] == "Confirmed"

def is_confirmed(booking):
    return booking["status"] == "Confirmed"
```

## Part 3 - Driving Function dan Rekursi

```python
def totals(datas, required):
    # Base case
    if not datas:
        return 0

    # Recursive case
    return datas[0][required] + totals(datas[1:], required)
```

## Part 4 - Pipeline

```python
#202410370110424
#Skenario 1
confirmed_bookings = list(filter(is_confirmed, bookings))
total_price_per_night = totals(confirmed_bookings, "price_per_night")
print(f"Total akumulasi Harga per Malam bookung confirmed: {total_price_per_night}")
#Skenario 2
taxed_bookings = list(map(apply_pricing, filter(is_confirmed, bookings)))
total_taxed_cost = totals(taxed_bookings, "total_cost")
print(f"Total cost confirm dengan pajak: {total_taxed_cost}")
```
