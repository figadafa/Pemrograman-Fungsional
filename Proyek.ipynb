```python
# 3. TUPLE MENU
menu = (
    "Login",
    "Registrasi",
    "Keluar",
    "Catat OTW",
    "Lihat Data",
    "Analisis Data",
    "Logout"
)
def get_string_input(prompt):
    while True:
        value = input(prompt).strip()
        if value:
            return value
        else:
            print("Input tidak boleh kosong. Mohon coba lagi.")

def get_numeric_input(prompt, type_func=int):
    while True:
        try:
            value_str = input(prompt).strip()
            value = type_func(value_str)

            if value < 0:
                print("Angka tidak boleh negatif. Mohon coba lagi.")
                continue

            return value

        except ValueError:
            print(
                f"Input tidak valid. Mohon masukkan angka yang benar "
                f"(misal: {type_func.__name__ == 'int' and '123' or '12.3'})."
            )
def get_time_input(prompt):
    while True:
        time_str = input(prompt).strip()
        try:
            datetime.datetime.strptime(time_str, "%H:%M")
            return time_str
        except ValueError:
            print(
                "Format waktu tidak valid. Mohon masukkan dalam format "
                "HH:MM (misal: 07:30)."
            )
```

```python
# 4. FUNGSI TAMPILKAN MENU

def validasi_login(accounts, user_id, password):
    return accounts.get(user_id) == password
def tampilkan_menu_program():
    print("\n" + "=" * 40)
    print(" SISTEM PENCATATAN EKSPERIMEN OTW")
    print("=" * 40)

    if current_user_id:
        print(f"Status: User Login ({current_user_id})")
        opsi_menu_saat_ini = menu[3:]
    else:
        print("Status: Guest")
        opsi_menu_saat_ini = menu[:3]

    for i, option in enumerate(opsi_menu_saat_ini, start=1):
        print(f"{i}. {option}")

    print("-" * 40)
    return len(opsi_menu_saat_ini)


def cari_akun(accounts, user_id):
    return user_id in accounts
```

```python
# 6.FUNGSI LOGIN

def login():
    global current_user_id

    print("\n=== LOGIN ===")
    user_id = get_string_input("Masukkan ID pengguna: ")
    password = get_string_input("Masukkan password: ")

    if validasi_login(accounts, user_id, password):
        current_user_id = user_id
        print(f"Login berhasil. Selamat datang, {profiles[user_id]['nama']}!")
    else:
        print("Login gagal. ID pengguna atau password salah.")
```

```python
# 7. FUNGSI MANIPULASI DATA OTW

import copy
def make_predicate_by_owner(user_id):
  
    def predicate(record):
        return record["owner_id"] == user_id
    return predicate


def make_predicate_by_activity(aktivitas):
    def predicate(record):
        return record["aktivitas"].lower() == aktivitas.lower()
    return predicate


def filter_records_by_predicate(data, predicate):
    return list(filter(predicate, data))


def filter_otw_data(data, user_id):
    hasil_filter = filter_records_by_predicate(
        data, make_predicate_by_owner(user_id)
    )
    return [copy.deepcopy(record) for record in hasil_filter]


def make_getter(key):
    def getter(record):
        return record[key]
    return getter

def tambah(a, b):
    return a + b


def add_average_speed(record):
    durasi = record["durasi_menit"]
    kecepatan = record["jarak_km"] / (durasi / 60) if durasi > 0 else 0.0
    return {**record, "kecepatan_km_jam": round(kecepatan, 2)}


def total_dengan_reduce(records, metric_key):
    return functools.reduce(tambah, map(make_getter(metric_key), records), 0)


def hitung_total_recursive(records, metric_key):
    if not records:
        return 0
    return records[0][metric_key] + hitung_total_recursive(
        records[1:], metric_key
    )

def _hitung_durasi_menit(waktu_otw_str, waktu_tiba_str):
    format_waktu = "%H:%M"

    waktu_otw = datetime.datetime.strptime(
        waktu_otw_str, format_waktu
    )
    waktu_tiba = datetime.datetime.strptime(
        waktu_tiba_str, format_waktu
    )

    durasi = waktu_tiba - waktu_otw

    # Menangani perjalanan yang melewati tengah malam
    if durasi.total_seconds() < 0:
        durasi += datetime.timedelta(days=1)

    return int(durasi.total_seconds() / 60)

def buat_catatan_otw(
    data, user_id, aktivitas,
    waktu_otw, waktu_tiba, jarak_km
):
    data_baru = copy.deepcopy(data)

    id_baru = max(
        (rec["id"] for rec in data_baru),
        default=0
    ) + 1

    catatan_baru = {
        "id": id_baru,
        "owner_id": user_id,
        "aktivitas": aktivitas,
        "waktu_otw": waktu_otw,
        "waktu_tiba": waktu_tiba,
        "durasi_menit": _hitung_durasi_menit(
            waktu_otw, waktu_tiba
        ),
        "jarak_km": jarak_km
    }

    return data_baru + [catatan_baru]

def catat_otw():
    global otw_data, next_record_id

    if not current_user_id:
        print("Anda harus login untuk mencatat OTW.")
        return

    print("\n--- Catat Pernyataan OTW Baru ---")

    aktivitas = get_string_input(
        "Aktivitas (misal: Kuliah, Rapat): "
    )
    waktu_otw_str = get_time_input(
        "Waktu mengatakan OTW (HH:MM, misal: 07:30): "
    )
    waktu_tiba_str = get_time_input(
        "Waktu tiba di tujuan (HH:MM, misal: 07:47): "
    )
    jarak_km = get_numeric_input(
        "Jarak (dalam KM, misal: 4.5): ",
        type_func=float
    )

    otw_data = buat_catatan_otw(
        otw_data,
        current_user_id,
        aktivitas,
        waktu_otw_str,
        waktu_tiba_str,
        jarak_km
    )

    next_record_id = max(
        (rec["id"] for rec in otw_data),
        default=0
    ) + 1

    print(
        f"Catatan OTW dengan ID {next_record_id - 1} "
        "berhasil ditambahkan!"
    )
    print(
        f"Durasi perjalanan: "
        f"{_hitung_durasi_menit(waktu_otw_str, waktu_tiba_str)} menit"
    )
def lihat_data():
    if not current_user_id:
        print("Anda harus login untuk melihat data OTW.")
        return

    print("\n--- Riwayat OTW Saya ---")

    user_records = filter_otw_data(
        otw_data,
        current_user_id
    )

    if not user_records:
        print("Anda belum memiliki catatan OTW.")
        return

    for rec in user_records:
        print(f"ID Catatan     : {rec['id']}")
        print(f"Aktivitas      : {rec['aktivitas']}")
        print(f"Waktu OTW      : {rec['waktu_otw']}")
        print(f"Waktu Tiba     : {rec['waktu_tiba']}")
        print(f"Durasi         : {rec['durasi_menit']} menit")
        print(f"Jarak          : {rec['jarak_km']} KM")
        print("-" * 30)




```

```python
# 7b. FUNGSI ANALISIS DATA OTW

def analisis_data_otw(data, aktivitas=None):
    if aktivitas:
        records = filter_records_by_predicate(
            data, make_predicate_by_activity(aktivitas)
        )
    else:
        records = list(data)

    # map(): menambahkan kecepatan rata-rata ke setiap record
    records = list(map(add_average_speed, records))

    jumlah = len(records)
    if jumlah == 0:
        return {"jumlah_catatan": 0}

    total_durasi = hitung_total_recursive(records, "durasi_menit")  # recursion
    total_jarak = total_dengan_reduce(records, "jarak_km")          # reduce
    total_kecepatan = total_dengan_reduce(records, "kecepatan_km_jam")

    return {
        "jumlah_catatan": jumlah,
        "total_durasi_menit": total_durasi,
        "total_jarak_km": round(total_jarak, 2),
        "rata_rata_durasi_menit": round(total_durasi / jumlah, 1),
        "rata_rata_kecepatan_km_jam": round(total_kecepatan / jumlah, 2),
    }


def analisis_data():
    if not current_user_id:
        print("Anda harus login untuk menganalisis data OTW.")
        return

    print("\n--- Analisis Data OTW Saya ---")
    aktivitas = input(
        "Filter aktivitas (kosongkan untuk semua): "
    ).strip()

    data_user = filter_otw_data(otw_data, current_user_id)
    hasil = analisis_data_otw(data_user, aktivitas or None)

    if hasil["jumlah_catatan"] == 0:
        print("Tidak ada catatan OTW yang sesuai.")
        return

    print(f"Jumlah catatan             : {hasil['jumlah_catatan']}")
    print(f"Total durasi               : {hasil['total_durasi_menit']} menit")
    print(f"Total jarak                : {hasil['total_jarak_km']} KM")
    print(f"Rata-rata durasi           : {hasil['rata_rata_durasi_menit']} menit")
    print(f"Rata-rata kecepatan        : {hasil['rata_rata_kecepatan_km_jam']} km/jam")
```

```python
# 9. FUNGSI LOGOUT

def buat_pesan_logout(profiles, user_id):
    if not user_id:
        return "Tidak ada pengguna yang sedang login."

    nama = profiles.get(user_id, {}).get("nama", user_id)
    return f"Logout berhasil! Sampai jumpa, {nama}."


def logout():
    global current_user_id

    pesan = buat_pesan_logout(profiles, current_user_id)
    current_user_id = None

    print(pesan)
```

```python
# 10. MAIN LOOP PROGRAM
def tentukan_aksi_menu(pilihan, sedang_login):
    menu_guest = {
        1: "login",
        2: "registrasi",
        3: "keluar"
    }
    menu_user = {
        1: "catat_otw",
        2: "lihat_data",
        3: "analisis_data",
        4: "logout"
    }
    menu_aktif = menu_user if sedang_login else menu_guest
    return menu_aktif.get(pilihan, "tidak_valid")

def main():
    while True:
        num_options = tampilkan_menu_program()

        pilihan = get_numeric_input(
            f"Pilih menu (1-{num_options}): "
        )

        aksi = tentukan_aksi_menu(
            pilihan, sedang_login=bool(current_user_id)
        )

        if aksi == "login":
            login()
        elif aksi == "registrasi":
            registrasi()
        elif aksi == "keluar":
            print(
                "Terima kasih telah menggunakan sistem ini. "
                "Sampai jumpa!"
            )
            break
        elif aksi == "catat_otw":
            catat_otw()
        elif aksi == "lihat_data":
            lihat_data()
        elif aksi == "analisis_data":
            analisis_data()
        elif aksi == "logout":
            logout()
        else:
            print("Pilihan tidak valid. Mohon coba lagi.")

        input("\nTekan ENTER untuk melanjutkan...")


# Menjalankan program

if __name__ == "__main__":
    main()
```
